```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Exam_Score'],
    bins=20,
    kde=True
)

plt.title("Exam Score Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Hours_Studied",
    y="Exam_Score"
)

plt.title("Study Hours vs Exam Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Exam_Score"
)

plt.title("Attendance vs Exam Score")

plt.show()
```

```python
def performance(score):
    if score >= 85:
        return "Excellent"
    elif score >= 70:
        return "Good"
    elif score >= 50:
        return "Average"
    else:
        return "Poor"


df['Performance_Level'] = df['Exam_Score'].apply(performance)

df.head()
```

```python
df['Study_Category'] = pd.cut(
    df['Hours_Studied'],
    bins=[0,3,6,10],
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)

df.head()
```

```python
df.columns
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="Performance_Level"
)

plt.title("Student Performance Levels")

plt.show()
```

```python
df.to_csv(
    r"D:\github\Student Performance Analysis\output\cleaned_student_performance.csv",
    index=False
)

print("File saved successfully")
```

```python
import pandas as pd

result = pd.read_sql(
    "SELECT * FROM student_performance LIMIT 10",
    engine
)

result
```

```python
pd.read_sql(
"""
SELECT 
    AVG(Exam_Score) AS Avg_Score,
    Hours_Studied,
    Attendance
FROM student_performance
GROUP BY Hours_Studied, Attendance
ORDER BY Avg_Score DESC
LIMIT 10;
""",
engine
)
```

```python
pd.read_sql(
"""
SELECT
    CASE
        WHEN Attendance >= 90 THEN 'High Attendance'
        WHEN Attendance >= 75 THEN 'Medium Attendance'
        ELSE 'Low Attendance'
    END AS Attendance_Group,
    AVG(Exam_Score) AS Average_Score
FROM student_performance
GROUP BY Attendance_Group;
""",
engine
)
```

```python
pd.read_sql(
"""
SELECT
    Gender,
    COUNT(*) AS Students,
    AVG(Exam_Score) AS Average_Score
FROM student_performance
WHERE Exam_Score < 60
GROUP BY Gender;
""",
engine
)
```

```python
pd.read_sql(
"""
SELECT
    Parental_Education_Level,
    AVG(Exam_Score) AS Average_Score
FROM student_performance
GROUP BY Parental_Education_Level
ORDER BY Average_Score DESC;
""",
engine
)
```

```python
pd.read_sql(
"""
SELECT
    Motivation_Level,
    AVG(Exam_Score) AS Average_Score,
    COUNT(*) AS Students
FROM student_performance
GROUP BY Motivation_Level
ORDER BY Average_Score DESC;
""",
engine
)
```
